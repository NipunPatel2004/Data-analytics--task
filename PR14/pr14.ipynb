```python
# PR_14: Exploratory Data Analysis (EDA)


# 1. Import Required Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Load Dataset
# Replace file path if needed
df = pd.read_csv("student_performance_data.csv")

# 3. INITIAL DATA INSPECTION

print("\n--- First 5 Rows ---")
print(df.head())  # View top records

print("\n--- Last 5 Rows ---")
print(df.tail())  # View bottom records

print("\n--- Dataset Shape ---")
print(df.shape)   # Rows and Columns

print("\n--- Column Names ---")
print(df.columns)

print("\n--- Data Info ---")
print(df.info())  # Data types + null values

print("\n--- Missing Values ---")
print(df.isnull().sum())  # Count missing values


# 4. DATA CLEANING

# Remove unwanted columns (index columns)
df = df.drop(["Unnamed: 0", "Unnamed: 0.1", "Unnamed: 0.2"], axis=1, errors='ignore')

print("\n--- Columns After Cleaning ---")
print(df.columns)

# 5. DESCRIPTIVE STATISTICS

print("\n--- Summary Statistics ---")
print(df.describe())  # Mean, Std, Min, Max

print("\n--- Value Counts (Categorical Data) ---")
print(df['Gender'].value_counts())
print(df['PartTimeJob'].value_counts())


# 6. UNIVARIATE ANALYSIS (Single Variable)

# Histogram for GPA
plt.figure()
plt.hist(df['GPA'])
plt.title("Distribution of GPA")
plt.xlabel("GPA")
plt.ylabel("Frequency")
plt.show()

# Histogram for Study Hours
plt.figure()
plt.hist(df['StudyHoursPerWeek'])
plt.title("Study Hours Distribution")
plt.xlabel("Hours")
plt.ylabel("Frequency")
plt.show()


# 7. BIVARIATE ANALYSIS (Two Variables)

# Scatter Plot: Study Hours vs GPA
plt.figure()
plt.scatter(df['StudyHoursPerWeek'], df['GPA'])
plt.xlabel("Study Hours Per Week")
plt.ylabel("GPA")
plt.title("Study Hours vs GPA")
plt.show()

# Scatter Plot: Attendance vs GPA
plt.figure()
plt.scatter(df['AttendanceRate'], df['GPA'])
plt.xlabel("Attendance Rate")
plt.ylabel("GPA")
plt.title("Attendance vs GPA")
plt.show()

# 8. MULTIVARIATE ANALYSIS

# Correlation Heatmap
plt.figure()
corr = df.corr()
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()



# 9. FINAL OBSERVATIONS (PRINT)
print("\n--- FINAL INSIGHTS ---")
print("1. Dataset contains student academic and personal data.")
print("2. No major missing values found.")
print("3. GPA mostly ranges between 2.5 to 3.5.")
print("4. Higher study hours and attendance lead to better GPA.")
print("5. Correlation shows positive relation between study hours and GPA.")
```
