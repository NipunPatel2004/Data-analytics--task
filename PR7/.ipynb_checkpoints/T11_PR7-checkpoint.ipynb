```python
# PART 1: DATA ACCESS CONTROL
# Data access control restricts which parts of the dataset users can view

# STEP 1: Import pandas library
# Pandas is used for loading and managing the dataset
import pandas as pd

# STEP 2: Load the dataset from storage
# read_csv() loads the stored dataset into a dataframe
data = pd.read_csv("student_performance_data.csv")

# STEP 3: Define columns that a normal user is allowed to see
# This simulates restricted data access
allowed_columns = ['StudentID', 'Gender', 'Age', 'Major']

# STEP 4: Create a restricted dataset containing only allowed columns
restricted_data = data[allowed_columns]

# STEP 5: Display the restricted dataset
# Only authorized fields are visible
print(restricted_data.head())
```

```python
# PART 2: DATA PRIVACY PROTECTION
# Privacy protection ensures personal information is hidden or anonymized

# STEP 6: Create a copy of the dataset to protect original data
privacy_data = data.copy()

# STEP 7: Mask StudentID values
# This converts real IDs into anonymous IDs
privacy_data['StudentID'] = privacy_data['StudentID'].apply(lambda x: "ID_#@*" + str(x)[-3:])

# STEP 8: Display anonymized dataset
print(privacy_data.head())
```

```python
# PART 3: DATA SECURITY
# Security practices protect datasets from data loss or unauthorized changes

# STEP 9: Create a secure backup of the dataset
# to_csv() saves the dataset to another file
#--CODE data.to_csv("secure_student_backup.csv", index=False)

# STEP 10: Display confirmation message
print("Secure dataset backup created successfully")
```

```python
# PART 4: DATA QUALITY STANDARDS
# Data quality checks ensure the dataset is clean and reliable

# STEP 11: Check for missing values in each column
missing_values = data.isnull().sum()

# Display missing values count
print("Missing Values in Dataset:")
print(missing_values)

# STEP 12: Check duplicate records
duplicate_records = data.duplicated().sum()

# Display number of duplicate records
print("\nNumber of Duplicate Records:", duplicate_records)
```

```python
# PART 5: DATA COMPLIANCE MONITORING
# This step checks if the dataset follows defined governance policies

# STEP 13: Identify invalid attendance values
# The rule is: AttendanceRate must be between 0 and 100
# This line finds rows where AttendanceRate is less than 0 or greater than 100
invalid_attendance = data[(data['AttendanceRate'] < 0) | (data['AttendanceRate'] > 100)]

# STEP 14: Display records that violate the rule
# If no rows are returned, it means the dataset follows the policy
print("Invalid Attendance Records:")
print(invalid_attendance)
```
