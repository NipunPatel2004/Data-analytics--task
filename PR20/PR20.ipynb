```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load Dataset
df = pd.read_csv("student_performance_data.csv")

print("=== Step 1: Dataset ===")
print(df.head())

# Step 2: Central Tendency
# Concept:
# Central tendency shows the "center" of data

# Using GPA as main variable
mean_val = df['GPA'].mean()       # Average value
median_val = df['GPA'].median()  # Middle value
mode_val = df['GPA'].mode()[0]   # Most frequent value

print("\n=== Step 2: Central Tendency ===")
print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)

# Step 3: Dispersion
# Concept:
# Dispersion shows how much data is spread

range_val = df['GPA'].max() - df['GPA'].min()   # Difference between max & min
variance_val = df['GPA'].var()                  # Spread from mean
std_dev = df['GPA'].std()                      # Average distance from mean

print("\n=== Step 3: Dispersion ===")
print("Range:", range_val)
print("Variance:", variance_val)
print("Standard Deviation:", round(std_dev, 2))

# Step 4: Quartiles
# Concept:
# Quartiles divide data into 4 equal parts

quartiles = df['GPA'].quantile([0.25, 0.5, 0.75])

print("\n=== Step 4: Quartiles ===")
print(quartiles)

# Step 5: Box Plot (Visualization)
# Concept:
# Box plot shows median, quartiles, and outliers

plt.figure()

plt.boxplot(df['GPA'])

plt.title("Statistical Spread of GPA")
plt.xlabel("GPA")

plt.show()
```
