```python
import pandas as pd
import numpy as np

# Step 1: Load Dataset

df = pd.read_csv("student_performance_data.csv")

print("=== Step 1: Raw Data ===")
print(df.head())
print("\nMissing Values:\n", df.isnull().sum())


# Step 2: Remove Duplicates

# Concept:
# Duplicate records can affect analysis and give incorrect results
# So we remove duplicate rows and keep only the first occurrence

df = df.drop_duplicates()

print("\n=== Step 2: Duplicates Removed ===")
print("Shape:", df.shape)

# Step 3: Handle Missing Values
for col in df.columns:
    
    # If column is CATEGORICAL
    if df[col].dtype == 'object':
        
        # Concept:
        # Mode is used to fill missing values in categorical data
        # because it replaces missing values with the most frequent category
        
        df[col] = df[col].fillna(df[col].mode()[0])
        
        # Alternative Concept:
        # We can also use "Unknown" to avoid bias in some cases
        # df[col] = df[col].fillna("Unknown")
    
    # If column is NUMERICAL
    else:
        
        # Concept:
        # Mean is used to fill missing numerical values
        # because it maintains the average of the dataset
        
        df[col] = df[col].fillna(df[col].mean())

print("\n=== Step 3: Missing Values Handled ===")
print(df.isnull().sum())


# Step 4: Data Type Conversion
# Convert data into correct format so calculations can be performed

for col in df.columns:
    try:
        df[col] = pd.to_numeric(df[col])
    except:
        pass

print("\n=== Step 4: Data Types ===")
print(df.dtypes)

# Step 5: Outlier Handling (IQR Method)
# We use IQR method to cap values within a valid range

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = np.where(df[col] > upper, upper,
                       np.where(df[col] < lower, lower, df[col]))
print("\n Lower range : ",lower)
print("\n Upper range : ",upper)
print("\n=== Step 5: Outliers Handled ===")


# Step 6: Standardize Text
# Standardizing text ensures consistency (e.g., 'male' → 'Male')

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.title()

print("\n=== Step 6: Text Standardized ===")


# Step 7: Final Output
print("\n=== Final Cleaned Data ===")
print(df.head())


# Step 8: Save Cleaned Dataset
df.to_csv("cleaned_student_data.csv", index=False)

print("\n Cleaned dataset saved successfully!")
```
