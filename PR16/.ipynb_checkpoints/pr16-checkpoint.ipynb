```python
# PR_16: Transforming student.json to Structured Table

import pandas as pd

# Step 1: Load the JSON File

df = pd.read_json("student.json")

print("Step 1 Output (Original Data):")
print(df.head())

# Step 2: Convert into Semi-Structured Format
# (Creating nested structure from your data)

data = df.to_dict(orient='records')

print("\nStep 2 Output (JSON Records):")
print(data[:3])

# Step 3: Flattening the Data (Parsing Phase)


# Even though your JSON is already flat,
# we use json_normalize to follow PR_16 concept

#df_structured = pd.json_normalize(data)
df_structured = pd.io.json.json_normalize(data)

print("\nStep 3 Output (Flattened Data):")
print(df_structured.head())

# Step 4: Handling Semi-Structured Lists (Exploding)

# Create a list column to demonstrate explode (required in exam)
df_structured['Activity_List'] = df_structured[['PartTimeJob', 'ExtraCurricularActivities']].values.tolist()

# Explode the list column
df_final = df_structured.explode('Activity_List')

print("\nStep 4 Output (Fully Structured Data):")
print(df_final[['StudentID', 'Activity_List', 'GPA']].head())

# Step 5: Summarization & Interpretation

# 1. Summary Statistics
print("\n--- Statistics ---")
print(df_final['GPA'].describe())

# 2. Average GPA per Activity
summary = df_final.groupby('Activity_List')['GPA'].mean()

print("\n--- Average GPA per Activity ---")
print(summary)

# Step 6: Export to CSV

df_final.to_csv("structured_output.csv", index=False)

print("\nFile 'structured_output.csv' is ready for Excel.")
```
