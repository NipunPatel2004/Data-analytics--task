```python
h = 5          # wall height
w = 7          # front house width
d = 4          # back room depth

door_h = 2
door_w = 2

# front window settings
win_row = 1
win_w = 2

# -------- ROOF --------
for i in range(h):
    # front triangle roof
    print(" " * (h - i - 1) + "*", end="")
    if i > 0:
        print(" " * (2 * i - 1) + "*", end="")

    # roof connection
    print("*", end="")

    # back slanted roof
    if i == 0:
        print("*" * (w + d))
    elif i == h - 1:
        print(" " * (w + d - 1) + "*")
    else:
        print(" " * (w + d) + "*")

# -------- ROOF BASE --------
print("*" * (2 * h) + "*" * (w + d))

# -------- WALLS (FRONT WINDOW + BACK DOOR) --------
for i in range(h):
    # -------- FRONT WALL --------
    if i == win_row:
        print("*" +
              " " * ((2 * h - 2 - win_w) // 2) +
              "*" * win_w +
              " " * ((2 * h - 2 - win_w) // 2) +
              "*", end="")
    else:
        print("*" + " " * (2 * h - 2) + "*", end="")

    # -------- BACK WALL --------
    if i >= h - door_h:
        # door (UNCHANGED)
        print("*" +
              " " * ((w + d - door_w) // 2) +
              "*" * door_w +
              " " * ((w + d - door_w) // 2 - 1) +
              "*")
    else:
        print("*" + " " * (w + d - 2) + "*")

# -------- BASE --------
print("*" * (2 * h) + "*" * (w + d))
```

```python
# ---------------- TOP ROOF ----------------
# We want odd number of stars: 1, 3, 5, 7, 9
# Roof width is 9 stars maximum

for i in range(1, 10, 2):

    total_width = 9                 # total roof width
    star_count = i                  # stars in current line

    # calculate spaces before stars
    space_count = (total_width - star_count) // 2

    # print spaces
    print(" " * space_count, end="")

    # print roof stars
    print("*" * star_count, end="")

    # space between roof and right block
    print(" ", end="")

    # print right side block
    print("*" * 16)


# ---------------- STRAIGHT LINE ----------------
# This line is printed only once

for i in range(1):

    left_part = "*" * 9
    gap = " "
    right_part = "*" * 16

    print(left_part + gap + right_part)


# ---------------- WINDOW LINE ----------------
# Only one window is required

for i in range(1):

    left_wall = "**"
    window = " |##|\n   |##|"
    right_wall = "**"
    gap = " "
    side_block = "*" * 16

    print(left_wall + window + right_wall + gap + side_block)


# ---------------- DOOR (2 LINES) ----------------
# Door height is 2 lines

for i in range(2):

    left_wall = "**"
    door_space = " " * 5
    right_wall = "**"
    gap = " "
    side_block = "*" * 16

    print(left_wall + door_space + right_wall + gap + side_block)


# ---------------- BOTTOM LINE ----------------
# Base of the structure (printed once)

for i in range(1):

    left_base = "*" * 9
    gap = " "
    right_base = "*" * 16

    print(left_base + gap + right_base)
```
