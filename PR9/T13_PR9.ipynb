```python
# STEP 1: Import required library
# Pandas is used for data acquisition and integration
# ---------------------------------------

import pandas as pd


# STEP 2: Data Acquisition from Source 1 (CSV file)
# Load academic dataset
# ---------------------------------

academic_df = pd.read_csv("academic_data.csv")

print("Academic Dataset:")
print(academic_df.head())


# STEP 3: Data Acquisition from Source 2 (JSON file)
# Load extra student dataset
# -----------------------------------

extra_df = pd.read_json("extra_student_data.json")

print("\nExtra Student Dataset:")
print(extra_df.head())


# STEP 4: Data Integration
# Merge both datasets using common key "StudentID"
# ---------------------------------------

integrated_dataset = pd.merge(academic_df, extra_df, on="StudentID", how="inner")



# STEP 5: Display integrated dataset
# ---------------------------------------

print("\nIntegrated Dataset:")
print(integrated_dataset.head())


# STEP 6: Store the integrated dataset
# --------------------------------------

integrated_dataset.to_csv("final_integrated_student_data.csv", index=False)

print("\nIntegration completed successfully.")
```
