```python
# ==========================================
# PR_24
# Structured Ingestion & Preparation
# Aquaculture Weather Dataset
# ==========================================

# ---------- IMPORT LIBRARIES ----------

import pandas as pd
import matplotlib.pyplot as plt

# ---------- STEP 1: LOAD DATASET ----------

df = pd.read_csv("B2K weather station.csv")

print("\nFIRST 5 ROWS")
print(df.head())

# ---------- STEP 2: CONVERT DATETIME ----------

df['Data'] = pd.to_datetime(
    df['Data'],
    dayfirst=True,
    errors='coerce'
)

# Remove invalid datetime rows
df = df.dropna(subset=['Data'])

# ---------- STEP 3: SET DATETIME INDEX ----------

df = df.set_index('Data')

# Sort sequentially
df = df.sort_index()

print("\nDATASET INFO")
print(df.info())

# ---------- STEP 4: KEEP REQUIRED COLUMNS ----------

df = df[['Temperatura', 'Humidade']]

# ---------- STEP 5: VALIDATION PROTOCOL ----------

# STEP 5.1: MONOTONICITY CHECK

print("\nMONOTONICITY CHECK")

is_monotonic = df.index.is_monotonic_increasing

print("Sequentially Ordered?:", is_monotonic)

# STEP 5.2: FREQUENCY INSPECTION

print("\nFREQUENCY INSPECTION")

time_deltas = (
    df.index.to_series()
    .diff()
    .dt.total_seconds()
)

freq_std = time_deltas.std()

print("Frequency Variation StdDev:", freq_std)

# STEP 5.3: MISSING VALUE CHECK

print("\nMISSING VALUE CHECK")

missing_ratio = df.isnull().mean() * 100

print(missing_ratio)

# ---------- STEP 6: RESAMPLING ----------

# Create fixed 15-minute intervals
df_resampled = df.resample('15min').mean()

print("\nRESAMPLED DATA")
print(df_resampled.head(10))

# ---------- STEP 7: INTERPOLATION ----------

# Fill only small gaps
df_resampled['Temperatura'] = (
    df_resampled['Temperatura']
    .interpolate(method='time', limit=4)
)

df_resampled['Humidade'] = (
    df_resampled['Humidade']
    .interpolate(method='time', limit=4)
)

# Remove remaining large-gap NaN rows
df_resampled = df_resampled.dropna()

print("\nAFTER INTERPOLATION")
print(df_resampled.head(10))

# ---------- STEP 8: VERIFY 15-MINUTE INTERVAL ----------

print("\nTIME FREQUENCY")
print(df_resampled.index.freq)

# ---------- STEP 9: VISUALIZATION ----------

plt.figure(figsize=(12,5))

# Scatter-style plot avoids fake straight lines
plt.plot(
    df_resampled.index,
    df_resampled['Temperatura'],
    marker='.',
    linestyle='None'
)

plt.title("Prepared Temperature Data")
plt.xlabel("Datetime")
plt.ylabel("Temperature")

plt.show()

# ---------- STEP 10: SAVE AS PARQUET ----------

df_resampled.to_parquet(
    "prepared_aquaculture_data.parquet"
)

print("\nPARQUET FILE SAVED SUCCESSFULLY")

# ---------- STEP 11: FINAL OUTPUT ----------

print("\nPRACTICAL COMPLETED SUCCESSFULLY")
```
