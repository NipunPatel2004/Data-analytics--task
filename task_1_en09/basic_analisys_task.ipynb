```python
from IPython.display import display
import pandas as pd

df = pd.read_csv("StudentPerformanceFactors.csv")

styled_df = df.head().style.set_properties(
    **{"text-align": "center"}
)
display(styled_df)
```

```python
import matplotlib.pyplot as plt


# Example: your dataset
# df = pd.read_csv("data.csv")  # Uncomment if using CSV

# Calculate average exam score by gender
avg_scores = df.groupby("Gender")["Exam_Score"].mean()

# Bar chart
plt.figure(figsize=(6,4))
bars = plt.bar(avg_scores.index, avg_scores.values, color=['skyblue' if v == avg_scores.max() else 'lightgray' for v in avg_scores.values])

# Add value labels on top of bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, round(yval,1), ha='center', fontsize=12 ,color="black")

plt.title("Average Exam Score by Gender",color="white")
plt.ylabel("Average Exam Score",color="white")
plt.ylim(0, max(avg_scores.values)+10)
plt.show()
```
