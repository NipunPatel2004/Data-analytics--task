```python
# PR_15: Structured Analysis of Tabular Data

# 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# 2. Load Dataset
df = pd.read_csv("student_performance_data.csv")

# STRUCTURED SUMMARIZATION

print("\n-----shape of dataste(row and col)----\n")
print(df.shape)

print("\n--- Summary Statistics ---")
print(df.describe())   

# CATEGORICAL INTERPRETATION

# Gender-wise average GPA
print("\n--- Gender-wise Average GPA ---")
print(df.groupby('Gender')['GPA'].mean())

# Major-wise average GPA
print("\n--- Major-wise Average GPA ---")
print(df.groupby('Major')['GPA'].mean())

# Part-time job vs GPA
print("\n--- Part-Time Job vs GPA ---")
print(df.groupby('PartTimeJob')['GPA'].mean())

# Extra Activities vs GPA
print("\n--- Activities vs GPA ---")
print(df.groupby('ExtraCurricularActivities')['GPA'].mean())

# VISUAL INTERPRETATION

# Boxplot for GPA (to detect outliers)
plt.figure()
df.boxplot(column='GPA')
plt.title("Distribution of GPA")
plt.ylabel("GPA")
plt.show()

# Boxplot for Study Hours
plt.figure()
df.hist(column='StudyHoursPerWeek')
plt.title("Study Hours Distribution")
plt.ylabel("Hours")
plt.show()

# ADDITIONAL VISUAL ANALYSIS (EXTRA MARKS)

# Bar chart: Average GPA by Gender
df.groupby('Gender')['GPA'].mean().plot(kind='bar')
plt.title("Average GPA by Gender")
plt.ylabel("GPA")
plt.show()

# Bar chart: Average GPA by Major
df.groupby('Major')['GPA'].mean().plot(kind='bar')
plt.title("Average GPA by Major")
plt.ylabel("GPA")
plt.xticks(rotation=45)
plt.show()

# FINAL INTERPRETATION

print("\n--- FINAL INSIGHTS ---")
print("1. Dataset shows student academic performance.")
print("2. GPA varies based on study habits and activities.")
print("3. Students with higher study hours tend to have better GPA.")
print("4. Participation in activities and jobs may impact performance.")
print("5. Boxplots help identify outliers in GPA and study hours.")
```
