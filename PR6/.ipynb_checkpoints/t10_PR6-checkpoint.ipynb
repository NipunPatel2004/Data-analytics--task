```python
# =====================================================
# DATA INTEGRATION
# =====================================================

import pandas as pd

# Load original dataset
data = pd.read_csv("student_performance_data.csv")

# Create two smaller datasets (simulate different sources)

academic_data = data[['StudentID','GPA','Major']]

activity_data = data[['StudentID','ExtraCurricularActivities','PartTimeJob']]

# Display sample datasets
print("Academic Data Sample:")
print(academic_data.head())

print("\nActivity Data Sample:")
print(activity_data.head())

# Integrate datasets using StudentID
integrated_data = pd.merge(
    academic_data,
    activity_data,
    on="StudentID"
)

print("\nIntegrated Dataset:")
print(integrated_data.head())
```

```python
# =====================================================
# PATTERN DISCOVERY
# =====================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv("student_performance_data.csv")

# STEP 1: Statistical summary
print("Statistical Summary:")
print(data.describe())

# STEP 2: GPA distribution
plt.hist(data['GPA'], bins=10)

plt.title("GPA Distribution")
plt.xlabel("GPA")
plt.ylabel("Number of Students")

plt.show()

# STEP 3: Study Hours vs GPA
plt.scatter(data['StudyHoursPerWeek'], data['GPA'])

plt.title("Study Hours vs GPA")
plt.xlabel("Study Hours Per Week")
plt.ylabel("GPA")

plt.show()

# STEP 4: Attendance vs GPA
plt.scatter(data['AttendanceRate'], data['GPA'])

plt.title("Attendance Rate vs GPA")
plt.xlabel("Attendance Rate")
plt.ylabel("GPA")

plt.show()

# STEP 5: Correlation analysis
correlation_matrix = data.corr(numeric_only=True)

print("\nCorrelation Matrix:")
print(correlation_matrix)

# STEP 6: Heatmap visualization
sns.heatmap(correlation_matrix, annot=True)

plt.title("Feature Correlation Heatmap")

plt.show()
```
