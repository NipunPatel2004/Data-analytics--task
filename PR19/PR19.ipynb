```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Load Dataset
df = pd.read_csv("student_performance_data.csv")

print("=== Dataset Loaded ===")
print(df.head())


# Step 2: Line Plot (Trend)
# Concept:
# Line plot shows trend or pattern over a continuous variable

plt.figure()

df_sorted = df.sort_values(by='StudyHoursPerWeek')

plt.plot(df_sorted['StudyHoursPerWeek'], df_sorted['GPA'], marker='o')

plt.title("Trend: Study Hours vs GPA")
plt.xlabel("Study Hours Per Week")
plt.ylabel("GPA")

plt.show()

# Step 3: Bar Chart (Comparison)
# Concept:
# Bar chart is used to compare values between categories

avg_gpa_gender = df.groupby('Gender')['GPA'].mean()

plt.figure()

plt.bar(avg_gpa_gender.index, avg_gpa_gender.values)

plt.title("Comparison: Average GPA by Gender")
plt.xlabel("Gender")
plt.ylabel("Average GPA")

plt.show()

# Step 4: Scatter Plot (Relationship)
# Concept:
# Scatter plot shows relationship between two variables

plt.figure()

plt.scatter(df['StudyHoursPerWeek'], df['GPA'])

plt.title("Relationship: Study Hours vs GPA")
plt.xlabel("Study Hours")
plt.ylabel("GPA")

plt.show()

# Step 5: Box Plot (Outliers)
# Concept:
# Box plot is used to detect outliers and understand data distribution

plt.figure()

plt.boxplot(df['GPA'])

plt.title("GPA Distribution and Outliers")
plt.ylabel("GPA")

plt.show()

# Step 6: Extra Comparison (Major vs GPA)
# Concept:
# Additional comparison using bar chart for better analysis

avg_gpa_major = df.groupby('Major')['GPA'].mean()

plt.figure()

plt.bar(avg_gpa_major.index, avg_gpa_major.values)

plt.title("Comparison: Average GPA by Major")
plt.xlabel("Major")
plt.ylabel("Average GPA")

plt.xticks(rotation=45)

plt.show()
```
