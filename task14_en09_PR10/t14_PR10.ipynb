```python
# STEP 1: Import Required Libraries (Analytical Environment)
# -----------------------------------------

import pandas as pd
import matplotlib.pyplot as plt


# STEP 2: Load Dataset
# ------------------------------------------

df = pd.read_csv("student_performance_data.csv")

print("Dataset Loaded Successfully")
print(df.head())


# STEP 3: Standardize Text Format
# Remove spaces and convert text to lowercase
# --------------------------------------------

df["Gender"] = df["Gender"].astype(str).str.strip().str.lower()
df["PartTimeJob"] = df["PartTimeJob"].astype(str).str.strip().str.lower()
df["ExtraCurricularActivities"] = df["ExtraCurricularActivities"].astype(str).str.strip().str.lower()
df["Major"] = df["Major"].astype(str).str.strip().str.lower()


# STEP 4: Convert Categorical Data to Numerical Representation
# (Using map to avoid FutureWarning)
# ----------------------------------------------

df["Gender"] = df["Gender"].map({"male": 1, "female": 0})
df["PartTimeJob"] = df["PartTimeJob"].map({"yes": 1, "no": 0})
df["ExtraCurricularActivities"] = df["ExtraCurricularActivities"].map({"yes": 1, "no": 0})


# STEP 5: Ensure Correct Data Types for Analysis
# ---------------------------------------------

df["Age"] = df["Age"].astype(int)
df["StudyHoursPerWeek"] = df["StudyHoursPerWeek"].astype(int)
df["AttendanceRate"] = df["AttendanceRate"].astype(float)
df["GPA"] = df["GPA"].astype(float)


# STEP 6: Display Standardized Dataset
#-----------------------------------------------------

print("\nStandardized Dataset:")
print(df.head())


# STEP 7: Visualization 1
# Study Hours vs GPA (Scatter Plot)
# ---------------------------------------------------

plt.figure()
plt.scatter(df["StudyHoursPerWeek"], df["GPA"])
plt.title("Study Hours vs GPA")
plt.xlabel("Study Hours Per Week")
plt.ylabel("GPA")
plt.show()


# STEP 8: Visualization 2
# Gender Distribution (Bar Chart)
# -------------------------------------------------

plt.figure()
df["Gender"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender (1 = Male, 0 = Female)")
plt.ylabel("Count")
plt.show()


# STEP 9: Visualization 3
# Attendance Rate Distribution (Histogram)
#--------------------------------------------

plt.figure()
plt.hist(df["AttendanceRate"])
plt.title("Attendance Rate Distribution")
plt.xlabel("Attendance Rate")
plt.ylabel("Frequency")
plt.show()


# STEP 10: Save Standardized Dataset
# -----------------------------------------------------

df.to_csv("standardized_student_dataset.csv", index=False)

print("\nStandardized dataset saved successfully.")
```
