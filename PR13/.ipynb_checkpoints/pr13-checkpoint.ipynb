```python
# 13. DATA TRANSFORMATION & STANDARDIZATION

# Import required libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler

# Step 0: Load Dataset

df = pd.read_csv('student_performance_data.csv')

print("Original Dataset:\n")
print(df.head(2))


# Step 1: Encode Categorical Variables

# Label Encoding (for 'Major')
le = LabelEncoder()
df['Major'] = le.fit_transform(df['Major'])

# One-Hot Encoding (for 'Gender')
df = pd.get_dummies(df, columns=['Gender'], drop_first=True)

print("\nAfter Encoding:\n")
print(df.head(2))

# Step 2: Scale / Normalize Numerical Columns

# Select numerical columns
num_cols = ['Age', 'StudyHoursPerWeek', 'AttendanceRate', 'GPA']

# -------- Standard Scaling --------
scaler_std = StandardScaler()
df_std = df.copy()
df_std[num_cols] = scaler_std.fit_transform(df[num_cols])

print("\nAfter Standard Scaling:\n")
print(df_std.head(2))

# -------- Min-Max Scaling --------
scaler_mm = MinMaxScaler()
df_mm = df.copy()
df_mm[num_cols] = scaler_mm.fit_transform(df[num_cols])

print("\nAfter Min-Max Scaling:\n")
print(df_mm.head(2))


# Step 3: Subset Features for Analysis

df_subset = df[['Age', 'GPA', 'Major', 'Gender_Male']]

print("\nSubset Data:\n")
print(df_subset.head(2))


# Step 4: Split Dataset into Analytical Subsets

groups = df.groupby('Major')

for major, group in groups:
    print(f"\nMajor Group: {major}")
    print(group.head(2))
```
