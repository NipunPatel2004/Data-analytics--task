```python
# STEP 1: Import Required Libraries
# These libraries create the environment for systematic data evaluation
# ------------------------------

import pandas as pd          # Used for data manipulation and analysis
import numpy as np           # Used for numerical operations
import matplotlib.pyplot as plt   # Used for plotting graphs
import seaborn as sns        # Used for advanced data visualization


# STEP 2: Load the Dataset
# Load the student performance dataset into the analysis environment
#---------------------------------

df = pd.read_csv("student_performance_data.csv")

# Display first 5 rows to verify that the dataset is loaded correctly
print("First 5 Rows of Dataset:")
print(df.head())



# STEP 3: Evaluate Dataset Structure
# Check dataset size, columns, and data types
#---------------------------------

print("\nDataset Shape (Rows, Columns):")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())


# STEP 4: Check for Missing Values
# Ensures data quality before performing analysis
#-----------------------------------

print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# If missing values exist, remove them
df = df.dropna()
```

```python

# STEP 5: Data Transformation
# Convert categorical values into numeric values for evaluation
# -----------------

# Convert Gender
df["Gender"] = df["Gender"].replace({
    "Male": 1,
    "Female": 0
})

# Convert PartTimeJob
df["PartTimeJob"] = df["PartTimeJob"].replace({
    "Yes": 1,
    "No": 0
})

# Convert ExtraCurricularActivities
df["ExtraCurricularActivities"] = df["ExtraCurricularActivities"].replace({
    "Yes": 1,
    "No": 0
})

# Show transformed dataset
print("\nTransformed Dataset:")
print(df.head())



# STEP 6: Descriptive Statistical Evaluation
# Calculate statistical measures for numerical features
#-------------------

print("\nStatistical Summary:")
print(df.describe())
```

```python
# STEP 7: Evaluate Average Performance Metrics
# Compute important metrics from the dataset
# -----------------------

avg_gpa = df["GPA"].mean()
avg_study_hours = df["StudyHoursPerWeek"].mean()
avg_attendance = df["AttendanceRate"].mean()

print("\nAverage GPA:", avg_gpa)
print("Average Study Hours Per Week:", avg_study_hours)
print("Average Attendance Rate:", avg_attendance)


# STEP 8: Analyze Relationship Between Study Hours and GPA
# Visualization to evaluate how study hours affect GPA
#-----------------------------

plt.scatter(df["StudyHoursPerWeek"], df["GPA"])
plt.xlabel("Study Hours Per Week")
plt.ylabel("GPA")
plt.title("Relationship Between Study Hours and GPA")
plt.show()

```

```python
# STEP 9: Group-Based Evaluation
# Analyze performance across different categories
# ----------------------------

print("\nAverage GPA by Gender:")
print(df.groupby("Gender")["GPA"].mean())

print("\nAverage GPA by Major:")
print(df.groupby("Major")["GPA"].mean())


# STEP 10: Correlation Analysis
# Evaluate relationships between numerical variables
# -------------------------------

correlation_matrix = df.corr(numeric_only=True)

print("\nCorrelation Matrix:")
print(correlation_matrix)

# Visualize correlation matrix
sns.heatmap(correlation_matrix, annot=True)

plt.title("Correlation Between Variables")
plt.show()
```

```python
# STEP 11: Save Processed Dataset
# Save the cleaned and transformed dataset for further analysis
#--------------------------------

df.to_csv("processed_student_data.csv", index=False)

print("\nProcessed dataset saved successfully.")
```
