```python
import pandas as pd
import numpy as np

# Step 1: Load Dataset
df = pd.read_csv("student_performance_data.csv")

print("=== Step 1: Original Dataset ===")
print(df.head())

# Step 2: Enhancement (Feature Engineering)

# Concept:
# Feature Engineering means creating new useful columns
# from existing data to improve analysis

# Create Performance Score (convert GPA to scale of 100)
df['Performance_Score'] = df['GPA'] * 25

# Create Study Efficiency (GPA per study hour)
df['Study_Efficiency'] = df['GPA'] / df['StudyHoursPerWeek']

print("\n=== Step 2: Enhanced Data ===")
print(df[['GPA', 'Performance_Score', 'Study_Efficiency']].head())

# Step 3: Scaling (Normalization)

# Concept:
# Scaling brings data into a common range (0 to 1)
# so that large values do not dominate analysis

df['GPA_Scaled'] = (df['GPA'] - df['GPA'].min()) / (df['GPA'].max() - df['GPA'].min())

print("\n=== Step 3: Scaled Data ===")
print(df[['GPA', 'GPA_Scaled']].head())

# Step 4: Restructuring (Pivot Table)

# Concept:
# Pivot table converts long format into wide format
# for better comparison and understanding

# Example: Average GPA by Gender and Major
pivot_df = df.pivot_table(values='GPA', index='Gender', columns='Major', aggfunc='mean')

print("\n=== Step 4: Pivot Table ===")
print(pivot_df)

# Step 5: Aggregation (Summary)
# Concept:
# Aggregation summarizes data to show overall performance

summary = df.groupby('Gender').agg({
    'GPA': 'mean',
    'StudyHoursPerWeek': 'mean',
    'AttendanceRate': 'mean'
}).rename(columns={
    'GPA': 'Avg_GPA',
    'StudyHoursPerWeek': 'Avg_StudyHours',
    'AttendanceRate': 'Avg_Attendance'
})

print("\n=== Step 5: Summary Table ===")
print(summary)

# Step 6: Save Enhanced Dataset
#df.to_csv("enhanced_student_data.csv", index=False)

print("\n Enhanced dataset saved successfully!")
```
